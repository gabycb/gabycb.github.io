##Audio Information
For the bias bounty, you will be provided pre-recorded emotional audio files. Each audio is  a minimum of 4.5s in length and in the .wav format. Each provided file represents one of the following emotions: happy, sad, angry, neutral. Each file's label is parsed from the first letter of the file name and corresponds to the first letter of it's emotion (ex: h_123456.wav is happy, n_78910.wav is neutral). The API outputs the main emotion detected, alongside the confidence of each emotion. We recommend using detected emotions above a 0.38 confidence.

Below, you will see a section to run files through the Valence emotion API singularly, and another section to run all files through the API and save the output in a spreadsheet.

### Run a Folder of Files

```python
AUDIO_FOLDER = "valence_audio"
```

```python
# TO DO: define output path for csv. Leave this as-is if you want it in whatever folder you're currently in
output_folder = ''
```

```python
# Generate a summary dataframe of each file and its emotion

data = []
for each in os.listdir(AUDIO_FOLDER):
  filepath = os.path.join(AUDIO_FOLDER, each)
  response = client.discrete.emotions(filepath)
  if each[0]=='h':
    true_emo = "happy"
  elif each[0]=='s':
    true_emo = "sad"
  elif each[0]=='a':
    true_emo = "angry"
  elif each[0]=='n':
    true_emo = "neutral"
  api_emotion = response["main_emotion"]
  confidence = response["confidence"]
  data.append([each, true_emo, api_emotion, confidence])
df = pd.DataFrame(data, columns=['filename', 'true_emotion', 'detected_emotion', 'confidence'])
df.to_csv(os.path.join(output_folder,"valence_output.csv"), index=False)
```
